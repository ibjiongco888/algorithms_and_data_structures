  # Question One: Check Duplicates in Tree

  Given the `root` of a binary tree, check whether it is contains a duplicate value. If a duplicate exists, return the duplicate value. If there are multiple duplicates, return the one with the closest distance to the root. If no duplicate exists, return -1.

  ## Examples

  Output: 2

  Input: `root = [1, 10, 2, 3, 10, 12, 12]`

  Output: 10

  ### Example 3

  ![](./images/q1_ex3.png)

  Input: `root = [10, 9, 8, 7]`

  Output: -1

#### Starter Code for Question 1

```python
# Definition for a binary tree node.
class TreeNode(object):
     def __init__(self, val = 0, left = None, right = None):
         self.val = val
         self.left = left
         self.right = right

from collections import deque

def is_duplicate(root: TreeNode) -> int:
    # If the tree is empty, there can be no duplicates.
    if not root:
        return -1
    
    # Set to keep track of values we have already seen:
    seen = set ()

    # Initialize a queue for BFS and start with the root node.
    q = deque([root])

    # Continue traversal while there are nodes in the queue.
    while q:
        # Remove the front node from the queue
        n = q.popleft()

        # If this node's value has already been seen,
        # it is the closest duplicate to the root.
        if n.val in seen:
            return n.val
        
        # Mark the current node's value as seen.
        seen.add(n.val)

        # Add the left child to the queue if it exists.
        if n.left:
            q.append(n.left)
        
        # Add the right child to the queue if it exists.
        if n.right:
            q.append(n.right)

    # If the entire tree is traversed without duplicates found:
    return -1    

 # Time and Space Complexity
 # Time: O(n) ==> visit each node once
 # Space: O(n) ==> set + queue       

 # I use BFS with a queue so the first duplicate encountered is guaranteed to be the closest to the root. 
 # A deque gives O(1) queue operations, making the solution O(n) time and space.
```

```python
# Validation and examples

# ------------------------------
# Sample test tree:
#         1
#        / \
#       2   3
#      /   /
#     4   2   <- duplicate

# Build the tree
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.right.left = TreeNode(2)  # duplicate

# Run the function
print(is_duplicate(root))  # Expected output: 2
```

```python
# ------------------------------
# Sample test tree:
#        5
#       / \
#      3   7
#     / \   \
#    2   4   8
#
# no duplicates

# Build the tree
root1 = TreeNode(5)
root1.left = TreeNode(3)
root1.right = TreeNode(7)
root1.left.left = TreeNode(2)
root1.left.right = TreeNode(4)
root1.right.right = TreeNode(8)

# Run the function
print(is_duplicate(root1))  # Expected output: -1
```

```python
# Your answer here

# My example:
# Input: s = “{([[[({})]]])}”
# Output: True

# My partner's example:
# Example 2 (invalid: bcuz altho the opening match with closing, but the order of closing is not last in first out. 
# s = "{[}]"
# Output: False

# Trace/walkthrough of the example that my partner made:
# The input string has these brackets "{[}]"". The last opening bracket is "[".
# The expected matching bracket to close it is "]", 
# but the next closing bracket is "}".
# The sequence is wrong; hence it is invalid.

# String: "{[}]"
# Step 1: { => opening bracket => push it
#           Stack: {
# Step 2: [ => another opening bracket => push it
#           Stack: {[
# Step 3: } => a closing bracket, but it must match the opening bracket at the top of
# the stack which is "[". This does not match, so this is when it becomes invalid.
```


-   Copy the solution your partner wrote. 


```python
# Your answer here

# Your answer here
def is_valid_brackets(s: str) -> bool:
    # match each type of closing bracket with the opening bracket of same type
    need_open = {')': '(', 
                 ']': '[', 
                 '}': '{'}
    stack = []

    for i in s:
        # If it's an opening bracket, put in in the stack
        if i in need_open.values():
            stack.append(i)

        # If it's a closing bracket, it must match the top
        else:
            if not stack or stack[-1] != need_open[i]:
                return False
            stack.pop()

    # valid when the stack is empty.
    return len(stack) == 0
```


-   Explain why their solution works in your own words.


```python
# Your answer here

# The solution works because it satisfies the conditions for a valid bracket sequence:
# 1. Opening brackets get pushed to a stack.
# 2. Closing brackets must match the most recent opening bracket.
# 3. Everything must match and the stack must end empty.

# I can explain it in detail by doing a walkthrough of the code:

# need_open tells us what each closing bracket should match.
# It uses stack[] to track opening brackets. Every time an opening bracket is seen, 
# it gets pushed to the stack. The top of the stack always represents the most recent bracket that has not been matched yet.
# When a closing bracket is seen, two things must be satisfied, and this part is checked in this line of the code:
#         if not stack or stack[-1] != need_open[i]:
#             return False
#         stack.pop()
# 1. The stack must not be empty (not stack). If there was any closing bracket, 
# but the stack is empty, this means that there is no opening bracket after all, 
# there is nothing to match, so the string is already invalid.
#         not stack = True, then return False
# 2. The top of the stack must match the correct opening bracket for the closing bracket that was just seen.
#          stack[-1] => the most recent opening bracket
#          need_open[i] => the opening bracket that should match the closing bracket i.
# So if:
#          stack[-1] != need_open[i], return False

# If both conditions 1 and 2 are met, we pop the stack because we found a valid pair.
#           stack.pop() 

# At the end of the stack, the stack must be empty:
#          return len(stack) == 0
# If anything left in the stack, it means there were opening brackets that were never closed. 
# It's placed at the end because leftover opening brackets can only be detected after processing the entire string.
```


-   Explain the problem’s time and space complexity in your own words.


```python
# Your answer here

# Time Complexity: O(n)
# The code goes through the string one character at a time, from start to finish. 
# If it’s an opening. It adds it to the stack. 
# If it’s a closing bracket, it checks the top of the stack and maybe removes it. 
# These actions (push, pop, compare) always takes the same amount of time no matter how long the string => O(1). 
# So, the total time depends on how many characters are in the string => n x O(1) = O(n).

# Space Complexity: O(n)
# In worst case, every character in the string is an opening bracket, so all of them get pushed to the stack. 
# That means the stack could grow to the same size as the input string, hence, the space complexity is O(n).
```
