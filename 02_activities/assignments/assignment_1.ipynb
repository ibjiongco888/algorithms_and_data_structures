## Question One: First Duplicate in List
**Description**  
Given a list of integers, return the **first value that appears more than once**. If there are multiple duplicates, return the one that appears **first** in the list. If no duplicate exists, return `-1`.

**Examples**
```python
Input: nums = [3, 1, 4, 2, 5, 1, 6]
Output: 1
```
```python
Input: nums = [7, 8, 9, 10]
Output: -1
```
```python
Input: nums = [4, 5, 6, 4, 6]
Output: 4
```

**Question 1 Starter Code**

````python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()
    for n in nums:
        if n in seen:
            return n
        seen.add(n)
    return -1

````

````python
tests = [
    [3, 1, 4, 2, 5, 1, 6],
    [7, 8, 9, 10],
    [4, 5, 6, 4, 6]
]

for t in tests:
    print(first_duplicate(t))
````


-   Code the solution to your assigned problem in Python (code chunk). Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Using that try to find the best time and space complexity solution!


````python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()
    for n in nums:
        if n in seen:
            return n
        seen.add(n)
    return -1
````


-   Explain why your solution works


````python
# The function scans the list from left to right and examines each element
# until a duplicate is encountered, at which point it terminates and returns
# the first duplicate value.
# It keeps track of all the numbers seen so far in a set.
# Sets in Python are implemented using hash tables, so checking if a number is in a set is very fast (average O(1) time).
# As soon as the function encounters a number that is already in the set, it knows this number has appeared before.
# Since the scan is from left to right, the first number detected this way is the first duplicate whose second occurrence 
# appears the earliest in the list.
# If the function finishes scanning the list and finds no duplicate, then it will return -1.
````


-   Explain the problem’s time and space complexity


````python
# Time complexity: O(n)
# The list is scanned once from left to right.
# Each set lookup and insertion takes O(1) on the average.
# Therefore, each loop iteration is: O(1) + O(1) = O(1)
# The loop runs once per element, for a total of n iterations (worst case).
# Thus, the overall complexity is: n x O(1) = O(n)
# In the best-case scenario, a duplicate is found immediately near the start of the list, allowing the function to terminate in O(1) time.

# Space complexity: O(n)
# A set is used to store elements that have already been seen.
# In the worst case (no duplicates), all the n elements are stored in the set, giving a space complexity of O(n).
````


-   Explain the thinking to an alternative solution (no coding required, but a classmate reading this should be able to code it up based off your text)


````python
# One alternative approach is to compare each element in the list with all elements that come after it using a
# nested loop.

# The idea is to start at the beginning of the list and, for each element, check whether the same value appears again later in the list.
# If a match is found, that value is a duplicate. However, because the problem asks for the duplicate whose second occurrence 
# appears earliest, we must ensure we do not immediately return the first matching value we find.

# To do this correctly, we keep track of the earliest position where any second occurrence is found. 
# As we scan the list, when we find a duplicate, we record the index of its second occurrence if it is earlier than 
# any previously recorded duplicate.
# After all comparisons are complete, we return the value whose second occurrence has the smallest index.
# If no duplicates are found after checking all pairs, we return -1.

# This approach does not use any additional data structures, 
# but it requires comparing many pairs of values, making it less efficient than the hash-set solution.
# While this approach is correct, it has a time complexity of O(n²), 
# which makes it less efficient than the hash-based solution that runs in linear time.
````
